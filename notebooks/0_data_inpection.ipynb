```python
# ==========================================================
# Cell 4: Gaussian Nature Analysis
# ==========================================================
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load your data (adjust path if running locally)
df = pd.read_csv('../data/clustering_dataset.csv')

# Set the visual style
sns.set_theme(style="whitegrid")

# Create a figure with two subplots
fig, axes = plt.subplots(1, 2, figsize=(15, 6))

# Plot 1: Distribution of X
sns.histplot(df['x'], kde=True, color="skyblue", ax=axes[0], bins=40)
axes[0].set_title('Distribution of X: Multi-modal Peaks', fontsize=14)
axes[0].set_xlabel('X coordinate')
axes[0].set_ylabel('Frequency')

# Plot 2: Distribution of Y
sns.histplot(df['y'], kde=True, color="salmon", ax=axes[1], bins=40)
axes[1].set_title('Distribution of Y: Multi-modal Peaks', fontsize=14)
axes[1].set_xlabel('Y coordinate')
axes[1].set_ylabel('Frequency')

# Overall title explaining the "Why"
plt.suptitle('Gaussian Nature Analysis: Identifying Mixture of Gaussians', fontsize=16, y=1.05)

plt.tight_layout()
plt.show()
```
